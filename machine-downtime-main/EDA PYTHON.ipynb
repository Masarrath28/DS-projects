```python
df['Failure'] = df['Downtime'].apply(lambda x: 'Yes' if x == 'Machine_Failure' else 'No')
df
```

```python
overall_failures=df['Failure'].value_counts()
overall_failures
```

```python
overall_failures = df['Failure'].value_counts()
overall_failures.plot(kind='bar', color=['red', 'green'])
plt.xlabel('Failure')
plt.ylabel('Count')
plt.title('Overall Counts of Failure Categories')
```

```python
failures = df[df['Failure'] == 'Yes']
failures
```

```python
failure_counts = failures.groupby(['Machine_ID', 'Assembly_Line_No']).size().reset_index(name='Failure_Count')
failure_counts
```

```python
sns.barplot(x='Machine_ID', y='Failure_Count', hue='Assembly_Line_No', data=failure_counts, palette='Set2')
plt.xlabel('Machine_ID')
plt.ylabel('Failure Count')
plt.title('Failures by Machine_ID and Assembly_Line_No')
plt.show()
```

```python
from scipy.stats import chi2_contingency
```

```python
# Pairplot for numerical variables
sns.pairplot(df,hue='Failure')
plt.show()
```

```python
cross_tab = pd.crosstab(df['Machine_ID'], df['Downtime'])
cross_tab
```

```python
df.head()
```

```python
contingency_table = pd.crosstab(df['Cutting(kN)'], df['Downtime'])
chi2, p, _, _ = chi2_contingency(contingency_table)
print(f"Chi-Square Value: {chi2}, p-value: {p}")
## lesser p value and higher chi2 value conclude strong assoiciation between variables
```

```python
contingency_table = pd.crosstab(df['Torque(Nm)'], df['Downtime'])
chi2, p, _, _ = chi2_contingency(contingency_table)
print(f"Chi-Square Value: {chi2}, p-value: {p}")
```

```python
contingency_table = pd.crosstab(df['Voltage(volts)'], df['Downtime'])
chi2, p, _, _ = chi2_contingency(contingency_table)
print(f"Chi-Square Value: {chi2}, p-value: {p}")
##p value grater than 0.05 , no association between 2 variables
```

```python
contingency_table = pd.crosstab(df['Tool_Vibration(µm)'], df['Downtime'])
chi2, p, _, _ = chi2_contingency(contingency_table)
print(f"Chi-Square Value: {chi2}, p-value: {p}")
##p value grater than 0.05 , no association between 2 variables
```
